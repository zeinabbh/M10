```python
import scanpy as sc
import anndata as ad
import numpy as np
```

```python
adata1 = sc.read_h5ad("/rds/general/user/ztb25/home/PBMC_datasets/1/PBMC1_clustering.h5ad")
adata4 = sc.read_h5ad("/rds/general/user/ztb25/home/PBMC_datasets/4/PBMC4_clustering.h5ad")
adata5 = sc.read_h5ad("/rds/general/user/ztb25/home/PBMC_datasets/5/PBMC5_clustering.h5ad")
adata6 = sc.read_h5ad("/rds/general/user/ztb25/home/PBMC_datasets/6/PBMC6_clustering.h5ad")
```

```python

# fixes from integration.ipynb
adata5.obs["sample"] = adata5.obs["sample"].cat.codes + 1
adata4.var.index = adata4.var["gene_symbols"].astype(str)
adata4.var_names_make_unique()
adata6.obs["sample"] = adata6.obs["sample"].astype("category").cat.codes + 1
```

```python

adata = ad.concat(
    {"1": adata1, "4": adata4, "5": adata5, "6": adata6},
    label="dataset", index_unique="-", join="inner", merge="unique",
)
```

```python

adata.obs["dataset"] = adata.obs["dataset"].astype("str")
adata.obs["sample"] = adata.obs["sample"].astype("str")
adata.obs["sample_combined"] = adata.obs["dataset"] + "_" + adata.obs["sample"]
```

```python

adata.write("/rds/general/user/ztb25/home/PBMC_datasets/concatenated_adata.h5ad")
print(adata)
```
